## How to Read the Feature Table

`COEFFICIENT` gives each gene’s contribution to **predicted AUC** in the locked Ridge model: positive means higher expression is associated with higher predicted AUC, negative with lower. `SIGN` is `+1` or `-1`. This is predictive association, not causation; interpret higher AUC as defined in the primary notebook.

`ABS_COEFFICIENT` ignores direction and ranks features. Larger values indicate stronger influence within this fitted model, not biological importance. `COEFFICIENT_MEAN` and `COEFFICIENT_MEDIAN` summarize refits; `COEFFICIENT_STD` measures variability. `POSITIVE_FRACTION` and `NEGATIVE_FRACTION` show directional stability: 1.00 means every refit, around 0.5 suggests instability. YTHDF3 is strong positive and perfectly stable; TMEM11 is strong negative and perfectly stable; DHRS7B has relatively low variability; WDR13 is mostly positive but less perfectly stable. Ridge distributes weight over correlated features, so no coefficient establishes a solitary biological driver.


## Targeted Literature Review Method

For each preselected feature, record (1) the model result, (2) independent literature, and (3) a clearly labelled interpretation/hypothesis. Search from specific to broad: `GENE + Erlotinib`, `GENE + drug resistance/sensitivity`, `GENE + EGFR`, `GENE + pathway`, `GENE + NSCLC`, then `GENE + lung cancer`.

Classify evidence as **Direct**, **Target/pathway related**, **Cancer-context related**, **Indirect/plausible**, or **No clear connection identified**. The last is valid and should not be replaced with speculation.
